```python
#Habeous funtion - gives the no. of conflicting queens
def calculate_cost(state):
    cost = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            # Check if queens attack each other
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


def hill_climbing():
    #Initial Condition
    current = [3, 1, 2, 0]

    while True:
        current_cost = calculate_cost(current)

        print("\nCurrent State:", tuple(current))
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("Goal State Reached!")
            return

        best_neighbor = current.copy()
        best_cost = current_cost

        # Generate all neighbors by swapping pairs
        for i in range(4):
            for j in range(i + 1, 4):
                neighbor = current.copy()
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                #habeous function calculation for all neighbours to get the best cost
                neighbor_cost = calculate_cost(neighbor)

                print("Swap:", (i, j),
                      "| Neighbor:", tuple(neighbor),
                      "| Cost:", neighbor_cost)

                #finding min cost (best cost)
                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_neighbor = neighbor

        # Stop if no better neighbor exists
        if best_cost >= current_cost:
            print("No better neighbor. Stopping.")
            return

        current = best_neighbor
        print("Selected Neighbor:", tuple(current))


# Execute the algorithm
hill_climbing()
```
